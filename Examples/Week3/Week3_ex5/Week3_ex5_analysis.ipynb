# Week3_ex5_analysis - Lorentz Force on a Wire in a Uniform 1 T Field vs B·I·L

![Week3_ex5 result](Images/Week3_ex5_result_image.png)

A 500 mm copper wire (10 mm diameter, 12-sided) carries 100 A along x inside box_h. Tangential-H boundaries on four of the box's faces, plus zero tangential H on the two y faces, force a uniform H = 1/μ0 along y, so B = 1 T. The force is taken with the Lorentz method on line_cal only, the 100 mm middle section of the wire. Magnetostatic solve.

This is a light check. Nothing unexpected shows up, so there are no hypothesis steps. The force should be B·I·L, perpendicular to both the current (x) and the field (y). The sign depends on the terminal polarity and the boundary's u direction, so only the axis and the magnitude are predicted. The wire's own field pushes radially on itself and should cancel over the segment.

$$\mathbf{F} = I\,\mathbf{L}\times\mathbf{B},\qquad |\mathbf{F}| = B I L = 1\,\mathrm{T}\times 100\,\mathrm{A}\times 0.1\,\mathrm{m} = 10\,\mathrm{N}$$

```python
import numpy as np
import pandas as pd

# 1. Applied field, current, and the length of line_cal
B = 1.0
I_wire = 100.0
L_cal = 0.1
F_expected = B * I_wire * L_cal

# 2. Force on line_cal, read once from Force Table 1 in the AEDT GUI (Setup1 : LastAdaptive, line_length = 500mm)
#    the table shows Force_x in fN and Force_y in uN; converted to N here
F = {
    "Force_mag": 10.001173,
    "Force_x": -207.616667e-15,
    "Force_y": 469.493755e-6,
    "Force_z": 10.001173,
}

pd.DataFrame({
    "component": list(F),
    "FEM [N]": list(F.values()),
    "expected [N]": [F_expected, 0.0, 0.0, F_expected],
    "difference [% of BIL]": [(F[k] - e) / F_expected * 100
                              for k, e in zip(F, [F_expected, 0.0, 0.0, F_expected])],
})
```

## Conclusion

The force on the 100 mm segment is 10.0012 N, 0.012% above B·I·L, and it points along z, perpendicular to both the current and the applied field, as expected. The two off-axis components are what's left of the wire's own radial self-force. Force_x is at the 1e-13 N level, which is numerically zero, and Force_y is 0.47 mN, or 0.005% of the main force. The self-force should cancel exactly over a symmetric segment, and at this size the leftover is what you'd expect from a slightly asymmetric mesh around a 12-sided wire. No candidate is worth chasing.

The report puts `line_length` on the x axis, but no sweep is defined, so the table has a single row (line_length = 500 mm). That single point is all this check needs. Nothing here needs a closer look.
